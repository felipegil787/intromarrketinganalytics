```python
!wget "https://github.com/javierherrera1996/IntroMarketingAnalytics/raw/refs/heads/main/SegundoCorte/clientes_segmentos%20(1).csv" -O clientes_segmentos.csv

import pandas as pd
df = pd.read_csv("clientes_segmentos.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, x='edad', y='ingresos', size='ingresos',
    hue='plataforma', alpha=0.6, sizes=(50, 500))
plt.title('Edad vs. Ingresos por Plataforma')
plt.xlabel('Edad')
plt.ylabel('Ingresos (USD)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Crear columnas RFM
df['Recency'] = df['dias_desde_ultima']
df['Frequency'] = df['num_compras']
df['Monetary'] = df['promedio_ticket'] * df['num_compras']
```
